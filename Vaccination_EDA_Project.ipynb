## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Create a copy of the dataset
df_clean = df.copy()

# Remove the unwanted text value from ISO_3_CODE
df_clean = df_clean[df_clean['ISO_3_CODE'] != 'Created: 2025-02-01 16:10 UTC']

# Remove completely empty rows
df_clean = df_clean.dropna(how='all')

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Convert YEAR to integer
df_clean['YEAR'] = df_clean['YEAR'].astype('Int64')

# Fill missing categorical values with 'Unknown'
categorical_columns = df_clean.select_dtypes(include='object').columns
df_clean[categorical_columns] = df_clean[categorical_columns].fillna('Unknown')

# Fill missing numerical values with median
numerical_columns = df_clean.select_dtypes(include=['int64', 'float64']).columns
df_clean[numerical_columns] = df_clean[numerical_columns].fillna(
    df_clean[numerical_columns].median()
)

# Check the cleaned dataset
print("Shape after cleaning:", df_clean.shape)
print("Missing values:")
print(df_clean.isnull().sum())
````

### What all manipulations have you done and insights you found?

The dataset was cleaned by removing unwanted values, empty rows, and duplicate records. The `YEAR` column was converted into an appropriate integer format. Missing categorical values were replaced with **"Unknown"**, while missing numerical values were handled using the **median**. The unique-value analysis showed that `ISO_3_CODE` contains 214 unique values, `VACCINECODE` contains 86 unique values, `SCHEDULEROUNDS` contains 7 unique values, `TARGETPOP` contains 15 unique values, and `GEOAREA` contains 2 unique values. The `AGEADMINISTERED` column has 425 unique values, showing considerable variation in vaccination age groups. `SOURCECOMMENT` also has 1,433 unique values, indicating diverse source-related information.
